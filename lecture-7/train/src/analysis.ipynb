```python
import pandas as pd

df_rail = pd.read_csv("../data/summary_by_railway.csv")
df_hour = pd.read_csv("../data/summary_by_hour_jst.csv")

display(df_rail)
display(df_hour)
```

```python
import matplotlib.pyplot as plt

# 見やすいように路線名を短縮（Toei.以降だけ）
plot_df = df_rail.copy()
plot_df["railway_short"] = plot_df["railway"].str.replace("odpt.Railway:Toei.", "", regex=False)

plt.figure()
plt.bar(plot_df["railway_short"], plot_df["abnormal_rate"])
plt.xticks(rotation=30, ha="right")
plt.ylabel("abnormal_rate")
plt.title("Abnormal rate by railway (Toei)")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plot_h = df_hour.copy()
plot_h["hour_jst"] = plot_h["hour_jst"].astype(int)

plt.figure()
plt.plot(plot_h["hour_jst"], plot_h["abnormal_rate"], marker="o")
plt.xticks(plot_h["hour_jst"])
plt.xlabel("hour (JST)")
plt.ylabel("abnormal_rate")
plt.title("Abnormal rate by hour (JST)")
plt.tight_layout()
plt.show()
```
